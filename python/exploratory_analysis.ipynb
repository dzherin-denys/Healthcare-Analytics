# 2. Patient Analysis
## 2.1 Age Distribution

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    patients["age"],
    bins=20
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.show()
```

### Key Insight

The patient population is broadly distributed across age groups, with a higher concentration of patients aged 50 years and older.

The distribution confirms findings from the SQL exploratory analysis, where patients aged 65+ represented the largest age group and demonstrated higher healthcare utilization and mortality rates.

## 2.2 Gender Distribution

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=patients,
    x="gender"
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")

plt.show()
```

### Key Insight

The patient population is evenly distributed across gender groups.

Female patients account for approximately 50.3% of the population, while male patients account for approximately 49.7%.

This balanced distribution supports unbiased comparisons across gender-based analyses.

## 2.3 Healthcare Expenses by Age

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=patients,
    x="age",
    y="healthcare_expenses",
    alpha=0.5
)

plt.title("Healthcare Expenses by Age")
plt.xlabel("Age")
plt.ylabel("Healthcare Expenses")

plt.show()
```

### Key Insight

Healthcare expenses increase substantially with age.

Older patients exhibit both higher average expenses and greater variability in healthcare costs.

The results support the findings from SQL analysis, where patients aged 65+ demonstrated the highest healthcare utilization and healthcare expenditures.

## 2.4 Average Healthcare Expenses by Age Group

```python
patients["age_group"] = pd.cut(
    patients["age"],
    bins=[0, 18, 35, 50, 65, 120],
    labels=["0-17", "18-34", "35-49", "50-64", "65+"]
)

avg_expenses = (
    patients.groupby("age_group")["healthcare_expenses"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(10,6))

sns.barplot(
    data=avg_expenses,
    x="age_group",
    y="healthcare_expenses"
)

plt.title("Average Healthcare Expenses by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Healthcare Expenses")

plt.show()
```

### Key Insight

Average healthcare expenses increase consistently across age groups.

Patients aged 65+ incur the highest healthcare costs, with average expenses exceeding 330,000, compared to approximately 24,000 for patients under 18.

These findings suggest that age is a major driver of healthcare spending and resource utilization.

## 2.5 Mortality by Gender

```python
patients["mortality_flag"] = (
    patients["deathdate"].notna().astype(int)
)
```

```python
mortality_gender = (
    patients.groupby("gender")["mortality_flag"]
    .mean()
    .reset_index()
)

mortality_gender["mortality_flag"] *= 100

plt.figure(figsize=(8,5))

sns.barplot(
    data=mortality_gender,
    x="gender",
    y="mortality_flag"
)

plt.title("Mortality Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Mortality Rate (%)")

plt.show()
```

### Key Insight

Male patients exhibit a higher mortality rate than female patients.

The mortality rate among male patients is approximately 14.7%, compared to 12.1% for female patients.

This difference suggests that gender may be associated with variations in patient outcomes within the population.

## 2.6 Mortality by Age Group

```python
mortality_age = (
patients.groupby("age_group")["mortality_flag"]
.mean()
.reset_index()
)

mortality_age["mortality_flag"] *= 100

plt.figure(figsize=(10,6))

sns.barplot(
data=mortality_age,
x="age_group",
y="mortality_flag"
)

plt.title("Mortality Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Mortality Rate (%)")

plt.show()
```

### Key Insight

Mortality rates increase significantly with age.

Patients aged 65 and older exhibit a mortality rate exceeding 40%, compared to less than 1% among patients under 18.

The results indicate that age is the strongest predictor of patient outcomes within the dataset.

# 3. Healthcare Utilization Analysis

## 3.1 Encounter Types

```python
encounter_types = (
    encounters["encounter_class"]
    .value_counts()
    .reset_index()
)

encounter_types.columns = [
    "encounter_class",
    "encounter_count"
]

plt.figure(figsize=(10, 6))

sns.barplot(
    data=encounter_types,
    x="encounter_class",
    y="encounter_count"
)

plt.title("Encounter Types")
plt.xlabel("Encounter Type")
plt.ylabel("Number of Encounters")

plt.xticks(rotation=45)

plt.show()
```

### Key Insight

Ambulatory encounters represent the largest share of healthcare utilization, followed by wellness and outpatient visits.

Most patient interactions occur in non-emergency settings, highlighting the importance of routine care, preventive medicine, and outpatient services.

Emergency and inpatient encounters account for a relatively small proportion of total healthcare utilization.

## 3.2 Healthcare Utilization by Age Group

```python
encounters_by_age = (
    patients.groupby("age_group")
    .size()
    .reset_index(name="patient_count")
)

encounter_counts = (
    encounters.groupby("patient_id")
    .size()
    .reset_index(name="encounter_count")
)

utilization = patients.merge(
    encounter_counts,
    on="patient_id",
    how="left"
)

utilization["encounter_count"] = (
    utilization["encounter_count"]
    .fillna(0)
)

utilization_by_age = (
    utilization.groupby("age_group")["encounter_count"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(10,6))

sns.barplot(
    data=utilization_by_age,
    x="age_group",
    y="encounter_count"
)

plt.title("Average Encounters per Patient by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Number of Encounters")

plt.show()
```

### Key Insight

Healthcare utilization increases consistently across age groups.

Patients aged 65 and older record the highest average number of healthcare encounters, while patients under 18 demonstrate the lowest utilization levels.

The findings reinforce previous analyses indicating that age is a major driver of healthcare demand and resource utilization.

## 3.3 Length of Stay Analysis

```python
plt.figure(figsize=(10,6))

sns.histplot(
    encounters["length_of_stay_days"],
    bins=30
)

plt.title("Length of Stay Distribution")
plt.xlabel("Length of Stay (Days)")
plt.ylabel("Number of Encounters")

plt.show()
```

### Key Insight

The vast majority of healthcare encounters have very short durations, typically less than one day.

This pattern reflects the predominance of ambulatory, wellness, and outpatient services within the healthcare system.

Longer stays are relatively rare and are likely associated with inpatient or complex healthcare encounters.

# 4. Clinical Analysis

## 4.1 Top Diseases

```python
top_diseases = (
    conditions[
        conditions["condition_description"]
        .str.contains("(disorder)", regex=False, na=False)
    ]
    .groupby("condition_description")
    .size()
    .reset_index(name="condition_count")
    .sort_values(
        "condition_count",
        ascending=False
    )
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_diseases,
    x="condition_count",
    y="condition_description"
)

plt.title("Top 10 Diseases")
plt.xlabel("Number of Records")
plt.ylabel("Disease")

plt.show()
```

### Key Insight

Respiratory illnesses represent the most frequently recorded diseases within the analyzed sample.

Acute viral pharyngitis and acute bronchitis account for the largest number of disease records, while injuries such as fractures and concussions also appear among the most common conditions.

The findings suggest that respiratory conditions and trauma-related diagnoses contribute substantially to the observed disease burden.

## 4.2 Top Diseases in 65+ Patients

```python
conditions_65 = conditions.merge(
    patients[["patient_id", "age"]],
    on="patient_id",
    how="inner"
)

top_diseases_65 = (
    conditions_65[
        (conditions_65["age"] >= 65)
        &
        (
        conditions_65["condition_description"]
        .str.contains("(disorder)", regex=False, na=False)
        )
    ]
    .groupby("condition_description")
    .size()
    .reset_index(name="condition_count")
    .sort_values(
        "condition_count",
        ascending=False
    )
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_diseases_65,
    x="condition_count",
    y="condition_description"
)

plt.title("Top Diseases in Patients Aged 65+")
plt.xlabel("Number of Records")
plt.ylabel("Disease")

plt.show()
```

### Key Insight

Respiratory diseases remain highly prevalent among patients aged 65 and older.

In addition, chronic kidney disease appears frequently within the elderly population, highlighting the growing burden of chronic conditions among older patients.

The findings further support the observation that healthcare utilization and disease complexity increase with age.

# 5. Correlation Analysis

## 5.1 Correlation Heatmap

```python
correlation_df = patients[
    [
        "age",
        "healthcare_expenses",
        "healthcare_coverage",
        "mortality_flag"
    ]
]

corr_matrix = correlation_df.corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="Blues",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

### Key Insight

Correlation analysis demonstrates that age has the strongest relationship with healthcare expenses, healthcare coverage, and mortality outcomes.

The positive correlation between age and mortality further confirms previous findings that older patients experience substantially higher healthcare utilization, healthcare costs, and mortality rates.

These results identify age as the primary driver of healthcare utilization and patient outcomes within the analyzed population.
